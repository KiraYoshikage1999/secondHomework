##Task1

```python
arr1 = np.random.randint(0, 101, size=10)
print("Initial array:", arr1)

min_val = np.min(arr1)
max_val = np.max(arr1)
mean_val = np.mean(arr1)

print(f"Minimum value: {min_val}")
print(f"Maximum value: {max_val}")
print(f"Mean value: {mean_val:.2f}")


sorted_arr1 = np.sort(arr1)
print("Array after sorting:", sorted_arr1)
print()
```

##Task 2

```python
arr2 = np.random.randint(1, 51, size=(4, 4))
print("Original array (4x4):")
print(arr2)

row_sums = np.sum(arr2, axis=1)
col_sums = np.sum(arr2, axis=0)

total_mean = np.mean(arr2)

print("Row sums:", row_sums)
print("Column sums:", col_sums)
print(f"Overall mean value: {total_mean:.2f}")
print()
```

##Task4

```python
arr4 = np.random.rand(10, 10)
print("Original array (10x10):")
print(np.round(arr4, 3))  # Rounded to 3 decimal places for better readability


min_val_4 = np.min(arr4)
max_val_4 = np.max(arr4)


min_idx_flat = np.argmin(arr4)
max_idx_flat = np.argmax(arr4)

# Convert flat indices to 2D coordinates (row, column)
min_idx = np.unravel_index(min_idx_flat, arr4.shape)
max_idx = np.unravel_index(max_idx_flat, arr4.shape)

print(f"\nMinimum value: {min_val_4:.4f}, Index (row, column): {min_idx}")
print(f"Maximum value: {max_val_4:.4f}, Index (row, column): {max_idx}")
print()
```

##Task 5

```python

arr5 = np.random.randint(-10, 11, size=15)
print("Original array:", arr5)


positive_elements = arr5[arr5 > 0]


even_elements = arr5[arr5 % 2 == 0]

print("Positive elements only:", positive_elements)
print("Even elements only:", even_elements)
print()
```

##Task 6

```python
arr6_a = np.arange(1, 6)
arr6_b = np.arange(6, 11)

print("Array 1:", arr6_a)
print("Array 2:", arr6_b)

concat_1d = np.concatenate((arr6_a, arr6_b))
print("\n1. 1D concatenation (np.concatenate):")
print(concat_1d)


array_2x5 = np.vstack((arr6_a, arr6_b))
print("\n2. 2x5 2D array (np.vstack):")
print(array_2x5)


array_5x2 = np.column_stack((arr6_a, arr6_b))
print("\n3. 5x2 2D array (np.column_stack):")
print(array_5x2)
```
